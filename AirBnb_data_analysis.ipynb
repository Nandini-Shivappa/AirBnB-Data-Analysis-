## Data exploration

Explore the dataset to understand its structure, including the number of rows and columns, data types of each column, and summary statistics. Identify potential missing values and outliers.


**Reasoning**:
Explore the dataset's structure, data types, summary statistics, missing values, and potential outliers.



```python
# Check the shape of the DataFrame
print("Shape of the DataFrame:", df.shape)

# Get a concise summary of the DataFrame
print("\nDataFrame Info:")
df.info()

# Generate descriptive statistics for numerical features
print("\nDescriptive Statistics for Numerical Features:")
display(df.describe())

# Explore categorical features
print("\nValue Counts for Categorical Features:")
for col in df.select_dtypes(include=['object']).columns:
    print(f"\nColumn: {col}")
    display(df[col].value_counts())
```

## Data cleaning

### Subtask:
Clean the data by handling missing values and outliers.


**Reasoning**:
Impute missing values and handle outliers as per the instructions.



```python
import pandas as pd

# Missing Value Imputation
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
df['last_review'] = df['last_review'].fillna('1900-01-01')
df.dropna(subset=['name', 'host_name'], inplace=True)

# Outlier Handling for 'price'
price_99th = df['price'].quantile(0.99)
df['price'] = df['price'].clip(upper=price_99th)

# Outlier Handling for 'minimum_nights'
min_nights_99th = df['minimum_nights'].quantile(0.99)
min_nights_1st = df['minimum_nights'].quantile(0.01)
df['minimum_nights'] = df['minimum_nights'].clip(lower=min_nights_1st, upper=min_nights_99th)

# Outlier Handling for 'calculated_host_listings_count'
host_listings_99th = df['calculated_host_listings_count'].quantile(0.99)
host_listings_1st = df['calculated_host_listings_count'].quantile(0.01)
df['calculated_host_listings_count'] = df['calculated_host_listings_count'].clip(lower=host_listings_1st, upper=host_listings_99th)
```

## Data wrangling

### Subtask:
Convert categorical features to numerical representations and engineer new features.


**Reasoning**:
Convert categorical features to numerical representations and engineer new features as per the instructions.



```python
import pandas as pd

# One-hot encode 'neighbourhood_group' and 'room_type'
df = pd.get_dummies(df, columns=['neighbourhood_group', 'room_type'], drop_first=True)

# Label encode 'neighbourhood'
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['neighbourhood'] = le.fit_transform(df['neighbourhood'])

# Engineer 'price_per_person' feature
df['price_per_person'] = df['price'] / df['minimum_nights']
```

## Data analysis

### Subtask:
Perform exploratory data analysis (EDA) on the cleaned and wrangled dataset.


**Reasoning**:
Analyze the distribution of key numerical variables and investigate the relationships between these variables and the target variable 'price'.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Numerical variables to analyze
numerical_vars = ['price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month',
                  'calculated_host_listings_count', 'availability_365', 'price_per_person']

# Histograms and box plots for numerical variables
plt.figure(figsize=(15, 10))
for i, var in enumerate(numerical_vars):
    plt.subplot(3, 3, i + 1)
    sns.histplot(df[var], kde=True)
    plt.title(f'Distribution of {var}')

plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 10))
for i, var in enumerate(numerical_vars):
    plt.subplot(3, 3, i + 1)
    sns.boxplot(y=df[var])
    plt.title(f'Box Plot of {var}')

plt.tight_layout()
plt.show()

# Correlation matrix and scatter plots
correlation_matrix = df[numerical_vars].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Variables')
plt.show()

plt.figure(figsize=(15, 10))
for i, var in enumerate(numerical_vars):
    plt.subplot(3, 3, i + 1)
    sns.scatterplot(x=var, y='price', data=df)
    plt.title(f'Scatter Plot of {var} vs. Price')

plt.tight_layout()
plt.show()

# Descriptive Statistics
print(df[numerical_vars].describe())
```

## Data visualization

### Subtask:
Visualize the findings from the data analysis.


**Reasoning**:
Generate the visualizations based on the analysis from the previous step.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Numerical variables
numerical_vars = ['price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month',
                  'calculated_host_listings_count', 'availability_365', 'price_per_person']

# Histograms with KDEs
plt.figure(figsize=(15, 10))
for i, var in enumerate(numerical_vars):
    plt.subplot(3, 3, i + 1)
    sns.histplot(df[var], kde=True)
    plt.title(f'Distribution of {var}')
plt.tight_layout()
plt.show()

# Box plots for price across categorical features
categorical_vars = ['neighbourhood_group_Brooklyn', 'neighbourhood_group_Manhattan',
                    'neighbourhood_group_Queens', 'neighbourhood_group_Staten Island',
                    'room_type_Private room', 'room_type_Shared room', 'neighbourhood']
plt.figure(figsize=(20, 15))
for i, var in enumerate(categorical_vars):
    plt.subplot(3, 3, i + 1)
    sns.boxplot(x=var, y='price', data=df)
    plt.title(f'Price Distribution by {var}')
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Scatter plots of price vs. numerical variables
plt.figure(figsize=(15, 10))
for i, var in enumerate(numerical_vars):
    plt.subplot(3, 3, i + 1)
    sns.regplot(x=var, y='price', data=df)  # Added regression line
    plt.title(f'Scatter Plot of {var} vs. Price')
plt.tight_layout()
plt.show()

# Bar chart of average price per neighborhood
plt.figure(figsize=(12, 6))
neighborhood_prices = df.groupby('neighbourhood')['price'].mean().sort_values(ascending=False)
sns.barplot(x=neighborhood_prices.index[:20], y=neighborhood_prices.values[:20]) # Top 20 neighborhoods
plt.xticks(rotation=90)
plt.title('Average Price per Neighborhood (Top 20)')
plt.xlabel('Neighborhood')
plt.ylabel('Average Price')
plt.tight_layout()
plt.show()

# Correlation heatmap
correlation_matrix = df[numerical_vars].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
